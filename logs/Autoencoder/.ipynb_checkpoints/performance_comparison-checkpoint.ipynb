```python
import matplotlib.pyplot as plt
import pandas as pd

performance_data = [
    {"Model": "PCA",
        "MSE": 0.2254,
        "MAE": 0.3630,
        "R² Score": 0.7746},
    {"Model": "Vanilla autoencoder",
        "MSE": 0.2372,
        "MAE": 0.3737,
        "R² Score": 0.6209},
    { "Model": "Autoencoder\n (lr scheduler)",
        "MSE": 0.2448,
        "MAE": 0.3784,
        "R² Score": 0.6046},
    {"Model": "Autoencoder\n (learnable weighting)",
        "MSE": 0.2369,
        "MAE": 0.3731,
        "R² Score": 0.6201},
    {"Model": "Autoencoder\n(feature weighting \nloss function)",
        "MSE": 0.2366,
        "MAE": 0.3728,
        "R² Score": 0.6208},
    {"Model": "Autoencoder\n(attention mechanism)",
        "MSE": 0.2365,
        "MAE": 0.3730,
        "R² Score": 0.6202}
]


df = pd.DataFrame(performance_data)
fig, ax = plt.subplots(3, 1, figsize=(12, 20))
metrics = ['MSE', 'MAE', 'R² Score']
colors = ['coral', 'hotpink', 'indigo']

for i, metric in enumerate(metrics):
    ax[i].bar(df['Model'], df[metric], color=colors[i])
    ax[i].set_title(f'{metric}')
    ax[i].set_ylabel('Model')

plt.show()
```
